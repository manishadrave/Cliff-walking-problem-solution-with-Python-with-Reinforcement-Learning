```python
#4*12  grid is availabel = 48 cells 
#Actions = up, down , forward, backward (0-3)
#reward normal cell = -1 target cell = 0 , cliff _cell = -100
#state is 48 bc of 48 possible cells 
#after hitting the cliff agent returns to the start position 
```

```python
import gymnasium as gym
import numpy as np 
import random
```

```python
env = gym.make("CliffWalking-v1")

print(env.observation_space.n) #no of state 
print(env.action_space.n) #no of movement for agent 

starting_state , _ = env.reset() #the first state value
print(starting_state)
```

```python
#using epsilion greedy policy so epsilion is also defined 
gamma = 0.99
epsilon = 0.1
alpha = 0.5
episodes = 500 
```

```python
#q value is for understanding which state and action pair is the best from action spaceQ[a,s] Q table is total no of q values in full grid 
Q = np.zeros((48,4)) #will be replaced with the value as the agent learns
```

```python
#Epsilion Policy decides action based on state
def epsilon_greedy(state):
    if random.random() < epsilon :
        return env.action_space.sample() #agent will explore randomly 

    else:
        return np.argmax(Q[state]) #agent will exploit
```

```python
#there's no need to visualize the env in training phase only observe it in testing phase
#SARSA is on policy algo 
for episode in range(episodes):
    # render = (episodes % 50 == 0 )     (for rendering in training)
    # if render:
    #      env = gym.make("CliffWalking-v1",render_mode = "human")
    # else:
    #   env = gym.make("CliffWalking-v1")
    #
    env = gym.make("CliffWalking-v1")

    done = False #when episode has ended 
    state, _ = env.reset()
    action = epsilon_greedy(state) 

    total_reward = 0
    episode_len = 0

    while not done:
        next_state,reward,terminated,truncated,_ = env.step(action)#for agent to take a action
        #next state is useful for next action
        done = terminated or truncated
        next_action = epsilon_greedy(next_state)

        #SARSA: Q vale is Update formula 
        Q[state, action] += alpha * ( reward + gamma *(Q[next_state, next_action] - Q[state, action]))

        state = next_state
        action = next_action

        total_reward += reward
        episode_len += 1

    print(f"Episode = {episode + 1}/500, total reward = {total_reward}, Episode len = {episode_len} ")
    env.close() #imp step 
```
